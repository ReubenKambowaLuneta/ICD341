```python
import numpy as np

x = np.array([10, 20, 30, 40, 50])
print(x)
print(x.shape)
print(x.ndim)
print(x.dtype)
```

```python
a = np.zeros(10)
b = np.ones(10)
c = np.arange(1, 11)
d = np.linspace(0, 1, 11)

print("a:", a)
print("b:", b)
print("c:", c)
print("d:", d)
```

```python
# Create the marks array from the lab
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

# 1. Display the first mark
print("1. First mark:", marks[0])

# 2. Display the last mark
print("2. Last mark:", marks[-1])

# 3. Display the first three marks
print("3. First three marks:", marks[:3])

# 4. Display marks from position 3 to 6
print("4. Marks from position 3 to 6:", marks[2:6])

# 5. Display every second mark
print("5. Every second mark:", marks[::2])
```

```python
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Std Dev:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Sum:", np.sum(marks))
```

```python
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print("Shape:", data.shape)
print("Dimensions:", data.ndim)
```

```python
print("1. First row:", data[0])
print("2. Second column:", data[:, 1])
print("3. Value 91:", data[1, 1])
print("4. First two rows:\n", data[:2])
print("5. Last two columns:\n", data[:, -2:])
```

```python
print("Mean per subject (columns):", np.mean(data, axis=0))
print("Mean per student (rows):", np.mean(data, axis=1))
```

```python
import matplotlib.pyplot as plt

# Generate 1,000 observations from a normal distribution
x = np.random.normal(loc=50, scale=10, size=1000)

# Calculate statistics
print("Mean:", np.mean(x))
print("Median:", np.median(x))
print("Standard Deviation:", np.std(x))
print("Minimum:", np.min(x))
print("Maximum:", np.max(x))

# Create a histogram
plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
